```python
import sys
sys.path.append("..")
from src.phonons import monatomic_dispersion

import numpy as np
import matplotlib.pyplot as plt

k = np.linspace(-np.pi, np.pi, 500) # first Brillouin zone (a = 1)
omega = monatomic_dispersion(k)
fig, ax = plt.subplots()
ax.plot(k, omega, label="chain")
ax.plot(k, np.abs(k), "--", label="sound wave v|k|") # v = a*sqrt(C/m) = 1
ax.set_xlabel("wavevector k [1/a]")
ax.set_ylabel("frequency ω [sqrt(C/m)]")
ax.set_ylim(0, 2.3)
ax.legend()
fig.savefig("../figures/01_monatomic.png", dpi=200)
plt.show()
```

```python
print(monatomic_dispersion(0, C=1, m=1, a=1))
print(monatomic_dispersion(np.pi, C=1, m=1, a=1))
print(monatomic_dispersion(np.pi, C=1, m=4, a=1))
print(monatomic_dispersion(np.pi, C=4, m=1, a=1))
print(monatomic_dispersion(0.01, C=1, m=1, a=1))
```
